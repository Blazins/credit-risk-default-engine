```python
import polars as pl
```

```python
file_path = "/home/mike/Documents/work/DS/projects/credit-risk/data/raw/accepted_2007_to_2018Q4.csv"
df_lazy = pl.scan_csv(file_path, infer_schema_length = 10000, schema_overrides={"id": pl.Utf8})
```

```python
#rows with text in the id column
bad_rows = df_lazy.filter(pl.col("id").str.contains("Total")).collect()
print(bad_rows)
```

```python
#Remove the bad rows
credit_risk = df_lazy.filter(~pl.col("id").str.contains("Total")).collect()
credit_risk.head(5)
```
